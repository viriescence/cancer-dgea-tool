For each gene: log2 fold change (mean tumor minus mean normal on log2 CPM), Welch's t-test, and Benjamini-Hochberg adjusted p-values. A gene is "up" or "down" at adjusted p < 0.05 and |log2 fold change| >= 1.

```python
%run python/dgea.py
```
